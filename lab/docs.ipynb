## Generate comments

```python
SLASH = '/'
ASTERISK = '*'
WIDTH = 79
BORDER = ASTERISK * 79
TOP_BORDER = SLASH + BORDER
BOTTOM_BORDER = BORDER + SLASH


def block_comment(s: str) -> str:
    """ Encase `s` in a block comment. """
    text_lines = s.split(NEWLINE)
    output_lines = [ TOP_BORDER, ASTERISK ]
    output_lines.extend([ ASTERISK + SPACE * 3 + line
                          for line in text_lines
                        ])
    output_lines.extend([ ASTERISK, BOTTOM_BORDER ])
    return NEWLINE.join(output_lines)
```

```python
COMMENT = """
Print the usual greeting.
"""
```

```python
print(rustdoc_comment(COMMENT, file_comment=False))
```

```python
comment = block_comment(COMMENT)
```

```python
DOT = '.'
MD_EXT = 'md'
DOCS_DIR_NAME = "docs"

def file2comment(s: str) -> str:
    """ Convert the file to an appropriate doc comment. """
    p = Path(DOCS_DIR_NAME) / (s + DOT + MD_EXT)
    assert(p.exists())
    FILES = lsd(p, output=False)
    return block_comment(p.read_text().strip())
```

```python
EXCLAMATION = '!'
def rustdoc_comment(s: str, file_comment=False) -> str:
    """ Convert the file text to a rustdoc style comment block. """
    return NEWLINE.join([SLASH * 2
                         + (EXCLAMATION if file_comment else SLASH)
                         + SPACE + string
                         for string in s.strip().split(NEWLINE)
                        ])
```
